```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = ["The cat sat on the mat",
            "The dog sat on the floor",
            "The cat and the dog played together"]
```

```python
documents[0]
```

```python
vectorizer = CountVectorizer()
```

```python
vectorizer
```

```python
X = vectorizer.fit_transform(documents)
```

```python
type(X)
```

```python
X.toarray()
```

```python
X_array = X.toarray()
```

```python
feature_names = vectorizer.get_feature_names_out()
```

```python
feature_names
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(X_array, columns=feature_names)
df
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [
    "The cat sat on the mat",
    "The dog sat on the floor",
    "The cat and the dog played together"
]
```

```python
tokenized_sentences = []
for sentence in sentences:
    tokenized_sentences.append(sentence.lower().split())

tokenized_sentences
```

```python
model = Word2Vec(sentences = tokenized_sentences, vector_size=100, min_count=1)
```

```python
model
```

```python
cat_embedding = model.wv['cat']
cat_embedding
```

```python
similar = model.wv.most_similar('cat', topn=5)
similar
```

```python
similarity = model.wv.similarity('cat', 'dog')
similarity
```
